## <u>Business Analytics & AI Insight Generation 

```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from IPython.display import display
```

```python
import os

for file in os.listdir():
    print(file)
```

```python
df = pd.read_csv("Sample - Superstore.csv", encoding='latin1')
df.head()
```

```python
df.info()
df.head()
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Year'] = df['Order Date'].dt.year

df['Month'] = df['Order Date'].dt.strftime('%b')

df['Month_Number'] = df['Order Date'].dt.month
```

```python
total_sales = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_orders = df['Order ID'].nunique()

avg_order = total_sales / total_orders

print("Total Sales:", round(total_sales,2))
print("Total Profit:", round(total_profit,2))
print("Total Orders:", total_orders)
print("Average Order Value:", round(avg_order,2))
```

#### Top region trend : 

```python
region_sales = (df.groupby('Region')['Sales'].sum().sort_values(ascending=False))

region_sales
```

```python
# finding best region:
top_region = region_sales.idxmax()

top_region_sales = region_sales.max()

print("Top Region:", top_region)

print("Sales:", round(top_region_sales,2))
```

#### Top Category trend:

```python
category_sales = (df.groupby('Category')['Sales'].sum().sort_values(ascending=False))
category_sales
```

```python
# finding best category:
top_category = category_sales.idxmax()

top_category_sales = category_sales.max()

print("Top Category:", top_category)

print("Sales:", round(top_category_sales,2))
```

#### Monthly Sales Trend 

```python

monthly_sales = (df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum().reset_index())
monthly_sales['Order Date'] = monthly_sales['Order Date'].astype(str)
monthly_sales.head()
```

```python
# Monthly sales visualization:
fig = px.line(
    monthly_sales,
    x='Order Date',
    y='Sales',
    title='Monthly Sales Trend'
)

fig.show()
```

```python
fig.write_image("Monthly Sales Trend.png")
```

#### Region Sales Chart:

```python
region_df = (df.groupby('Region')['Sales'].sum().reset_index())

fig = px.bar(
    region_df,
    x='Region',
    y='Sales',
    title='Sales by Region'
)

fig.show()
```

```python
fig.write_image("Sales by Region.png")
```

#### Category Sales Chart:

```python
category_df = (df.groupby('Category')['Sales'].sum().reset_index())

fig = px.bar(
    category_df,
    x='Category',
    y='Sales',
    title='Sales by Category'
)

fig.show()
```

```python
fig.write_image("Sales by Category.png")
```

#### Profit Analysis and Chart:

```python
region_profit = (df.groupby('Region')['Profit'].sum().reset_index())
fig = px.bar(
    region_profit,
    x='Region',
    y='Profit',
    title='Profit by Region'
)

fig.show()
```

```python
fig.write_image("Profit by Region.png")
```

#### Top 10 Product Chart

```python
top_products = (df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10).reset_index())

fig = px.bar(
    top_products,
    x='Sales',
    y='Product Name',
    orientation='h',
    title='Top 10 Products'
)

fig.show()
```

```python
fig.write_image("Top 10 Products.png")
```

## AI Insight Generation (Rule-Based)

```python
insights = []

insights.append(
    f"Total sales reached ${total_sales:,.0f}."
)

insights.append(
    f"Total profit reached ${total_profit:,.0f}."
)

insights.append(
    f"{top_region} was the highest revenue-generating region."
)

insights.append(
    f"{top_category} was the best-performing category."
)
```

#### Growth Analysis

```python
latest_month = monthly_sales.iloc[-1]['Sales']

previous_month = monthly_sales.iloc[-2]['Sales']

if latest_month > previous_month:
    
    insights.append(
        "Sales increased in the latest month."
    )
    
else:
    
    insights.append(
        "Sales decreased in the latest month."
    )
```

#### Business Recommendations

```python
recommendations = []

recommendations.append(
    f"Increase focus on {top_region} region due to strong sales performance."
)

recommendations.append(
    f"Expand inventory for {top_category} products."
)

recommendations.append(
    "Monitor low-profit regions for cost optimization."
)

recommendations.append(
    "Use monthly trend tracking for forecasting future demand."
)
```

```python
# Display Insights
print("AI GENERATED BUSINESS INSIGHTS")
print("-" * 50)

for item in insights:
    print("•", item)
```

```python
# Display Recommendations
print("BUSINESS RECOMMENDATIONS")
print("-" * 50)

for item in recommendations:
    print("•", item)
```

##### Export:-

```python
with open("insights_report.txt", "w") as f:

    f.write("BUSINESS ANALYTICS REPORT\n\n")

    f.write("KEY INSIGHTS\n")

    for item in insights:
        f.write(f"- {item}\n")

    f.write("\nRECOMMENDATIONS\n")

    for item in recommendations:
        f.write(f"- {item}\n")

print("Report Generated Successfully")
```
